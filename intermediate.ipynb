### Inheritance

### Multi Threading 

```python
import threading
import time

def hello(delay):
    time.sleep(delay)
    print("hello")

def naam (name,delay):
    time.sleep(delay)
    print(name)

t1= threading.Thread(target=hello,args=(0.5,))
t2= threading.Thread(target=naam,args=("Ikraama",0.4999))

t1.start()  #starts 
t2.start()  #starts in parallel 
# t1.join()   # Main thread will now wait until t1 finishes
# t2.join()   #then waits for t2 (if not already finished)
```

**A threading.Lock()** is used to *synchronize access* to the shared variable x. Both the *double()* and *halve()* functions acquire the lock before they begin modifying x, ensuring that only one thread can operate on x at a time. This prevents race conditions and ensures data consistency. The lock is explicitly acquired using lock.acquire() and released using lock.release() after the critical section (loop) finishes. This guarantees thread-safe operations on x.

```python
import threading
import time
 
x= 2000
lock = threading.Lock()

def double():
    global x,lock #take lock as global func
    lock.acquire()
    while x<4000:
        x*=2
        print(x)
        time.sleep(1)
    print("reached the maximum")
    lock.release()

def halve():
    global x,lock
    lock.acquire()
    while x >1:
        x//=2
        print(x)
        time.sleep(1)
    print("reached minimum")
    lock.release()


t1= threading.Thread(target=halve)
t2 = threading.Thread(target=double)


t1.start()
t2.start()
```

**Semaphores** used to limit the threade in use and allowing to use for limited amount of time to spend on main-thread 

```python
import threading
import time

semaphores = threading.BoundedSemaphore(value=3)

def access(thread_number):
    print(f"{thread_number} trying to get acccess")
    semaphores.acquire()
    print(f"{thread_number} has gained the access")
    time.sleep(10)
    print(f"{thread_number} released the access\n")
    semaphores.release()

for thead_number in range (1,11):
    t= threading.Thread(target=access,args=(thead_number,))
    t.start()
    time.sleep(0.5)
```
