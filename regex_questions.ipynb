```python
import re
james = "hello","hell","hola","hole"
pattern = r"^h[A-Za-z]{3}o$"

for name in james:
    if re.findall(pattern, name):
        print(name)

#x will be the value that matche the pattern but not naming the pattern
x = re.findall(pattern,james)
print(x)

reg_re = [name for name in james if re.findall(pattern,name)]
print(reg_re)
```

🔎 Extract only the valid-looking email addresses (must have @, then a domain name, and then .com, .in, etc.)

```python
text = "My emails are: ikraama123@gmail.com, test.user@outlook.in, hello@world"
pattern = r"\w+@\w+\.(?:com|in)"  #✅ (?:...) is a non-capturing group, so it groups .com|.in properly, but doesn’t mess with output.

print(re.findall(pattern, text))
```

Condition:
🔎 Extract only dates in this format:

dd/mm/yyyy

Day and month must have two digits

Year must be 4 digits

Accept only / as separator (not -)

So:

✅ 18/07/2025

❌ 17-07-2025 → wrong separator

❌ 07/18/2025 → wrong order

❌ 18-7-2025 → wrong separator and single digit

❌ 18/7/25 → 1-digit month & 2-digit year



```python
text = "Today is 18/07/2025. Yesterday was 17-07-2025. Wrong formats: 07/18/2025, 2025/07/18, 18/7/25, 18-7-2025."

pattern= r"\d{2}/\d{2}/\d{4}"
print(re.findall(pattern,text))
```

Extract only valid Twitter-style usernames that follow these rules:

✅ Conditions:

Starts with @

Followed by:

One or more letters, numbers, or underscores

Dots (.) are also allowed inside the name

Should NOT allow:

Just @ alone

@@anything

Spaces after @

* Group (?<!@) negative look behind to reject 

```python
text = "Follow us at @ikraama, @dev_guru123, and @AI.expert! Avoid spam like @ and @@invalid."
pattern = r"(?<!@)@\w+[\w\.]\w*"

print(re.findall(pattern,text))
```

```python
text = "Call us at +91-9876543210 or 9123456789. Landlines like 011-2345678 or +1-202-555-0183 are invalid. Also invalid: 123456789, 0012345678, 09876543210"

pattern = r"(?:\+91[-]?)?[789]\d{9}"
print(re.findall(pattern,text))
```

| Part               | Meaning                   | Simple Example     | Explanation                                                |
|--------------------|---------------------------|---------------------|------------------------------------------------------------|
| (?:...)            | Non-capturing group       | (?:abc)             | Grouping without saving it separately                      |
| \+91               | Match the +91 literally   | +91                 | `\+` means literal `+`, not "one or more"                  |
| [-]?               | Optional hyphen -         | - or nothing        | It means dash is optional                                  |
| (?:\+91[-]?)?      | The whole part is optional| +91- or +91 or none | Matches if there’s a country code                          |
| [789]              | Must start with 7, 8, or 9| 9 or 7 or 8         | Only valid Indian mobile numbers start this way            |
| \d{9}              | Then 9 digits             | 123456789           | Total of 10 digits now with the first [789]                |
